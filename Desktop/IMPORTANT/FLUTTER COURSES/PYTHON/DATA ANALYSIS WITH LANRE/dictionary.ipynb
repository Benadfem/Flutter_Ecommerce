```python
"""a program that will indicate counting status. from start to mid and to end """

for index in range(0, 101, 5):
    if index == 0:
        print(f"This is the beginning {index}")
    elif index == 50:
        print(f"Mid point {index}")
    elif index == 100:
        print(f"The end {index}")
    else:
        print(index)
```
