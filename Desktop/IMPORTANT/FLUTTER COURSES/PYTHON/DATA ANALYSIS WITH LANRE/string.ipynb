```python
first_name = 'adedoyin'
last_name = 'adedara'

print("first name:", first_name)
print("Last name:", last_name)

"""
string concertination is the joining of string datatypes 
the concertication is done with the + symbol
"""
full_name= first_name + " " + last_name # this is string concertination 
print("the complete name is:",full_name)
"""string formatting by using f command """
full_name = f"{first_name} {last_name} " #this is the best way to concertinate string 
print(full_name)
```

```python
first_name = 'adedoyin'
last_name = 'adedara'

print("first name:", first_name.upper())
print("Last name:", last_name.upper())

print("first name:", first_name.title())
print("Last name:", last_name.title())
```

```python
"""a simple program that communicates with the user through the console
    write a program to write a simple program to calculate the perimeter of a rectangle
    the user should input the parameter for the rectangle 
"""
length = int(input("Enter the length"))
breadth = int(input("Enter the breadth"))

perimeter = 2 * ( length + breadth)

print(f"The Perimeter of the rectangle is {perimeter}")
```
