Super class : this is the parent class, ie. the class on the upper hierachy. for example a FATHER  of a child is the super 
of the child. 

Sub class: this is the child class. ie. the class on the lower rank compared to a Super class. Sub class inherits from a super. e.g 
a child resembling his FATHER. 

programmatically, we shapes. Shape is a general name for every shape. area, perimeter(circumference) 
    examples shapes rectangle, square, circle, triangle, parallelogram etc. 

```python
"""A program that build a class of shapes and implements inheritance to the shapes """
# Create a parent class
class Shape:
    def __init__(self,name):
        # now we have to reference or initialise the name of the shape 
        # the constructor method holds the attributes of class
        self.name = name #the name is refered to as attribute. therefore every property of a class is called attribute
    def area(self):# every method in a class must have a self parameter(default) 
        print("Let's calculate your area ")
    def perimeter(self):
        print("We have perimeter")
"""An object of a class is bringing the class blueprint to a reality. """
sh1 = Shape("Biggy")
#you can now call the name attribute from the class Shape
print(sh1.name)
# we can callt he method from the class of Shape
# anytime you call a method. you should put the () so you can reach the content of the method 
sh1.area()
```

```python
"""Let's create a class the implements inheritance from the class Shape"""
class Triangle(Shape):# this is the declaration of inheritance class 
    #the Shape in the bracket is the class being inherited. so called the Super class 
    # pass #to avoid syntax error use pass. 
    pass
    
tri = Triangle("Yellow Tri")
tri.name
tri.area()
```

```python
class Rectangle(Shape):
    pass
rec = Rectangle("Yellow Rec")
print(rec.name)
print(rec.area())
print(rec.perimeter())
```

```python
"""Let's create a class the implements inheritance from the class Shape"""
class Triangle(Shape):# this is the declaration of inheritance class 
    #the Shape in the bracket is the class being inherited. so called the Super class 
    # pass #to avoid syntax error use pass. 
    def area(self,base, height): # Method Overriding 
        """Method overriding is the act of creating same method name that has been created before
        either through super class or in the early line of code """
        area = 0.5 * base * height
        return area
    def Perimeter(self, base, height):
        perimeter = base + height 
        return perimeter
    
tri = Triangle("Right angled Triangle")
print(f"The area of {tri.name} is: {tri.area(5, 9)}")

tri2 = Triangle("Equilateral Triangle")
print(f"The area of {tri2.name} is: {tri2.area(5, 5)}")
```

ENCAPSULATION: this means protecting the data in the class 
scope: the area or range at which data or variable can be seen, accessed or used in a program
two types of scope: 
global variable: can be seen at everywhere in the program 
local variable: local can only be seen in a function body that it's being created 

Encpalsulation needs authorization of data before it can use used or accessed. either globally or locally. 
meaning you cannot use a protect data outside the class that the data is being created. 

ENCAPULSATION is detnoted in python programming language by double_underscore sign (__)

now for you to access a protect data, you need the getter() and setter method() 

```python
class School:
    def __init__(self, name):
        self.name = name
    def subject(self, name):
        pass 

sch1 = School("Goodluck")
print(sch1.name)
```

```python
"""we will effect ancapsulation in the code below so the school name and methods willnot
be easily accessed by random call."""
class School:
    def __init__(self, name):
        #add __ to name variable below to encapulsate name
        self.__name = name 
    def subject(self, name):
        self.name = name 
        return name
    # the method below is the getter method that gives access to the protect data 
    def get_name(self):
        return self.__name

sch1 = School("Goodluck")
 # print(sch1.name)# there will be an attributeError when you access error message. 
# we will need to create a method to get the info from the class because it is protected. 
# print(sch1.get_name())
print(sch1.subject("Mathematics"))
```
